### Petal Length Distribution

The distribution of petal lengths for all 150 flowers is examined using a histogram.

```python
# Selecting the petal length values for all flowers.

petal_lengths = iris["petal_length"]
```

```python
# Calculating the mean and standard deviation of all petal lengths.

petal_mean = petal_lengths.mean()
petal_std = petal_lengths.std()

petal_mean, petal_std
```

The mean petal length across all 150 flowers is approximately 3.76 cm, while the standard deviation is approximately 1.77 cm. This means that there is quite a large amount of variation in petal length across the dataset, with many values differing noticeably from the overall mean. This variation may be influenced by the dataset containing three different Iris species.

### Preparing the Normal Distribution

To plot a normal distribution over the histogram, a range of values is needed for the x-axis. The `np.linspace()` function is used to create 100 evenly spaced values between the minimum and maximum petal lengths. These values will be used to calculate points along the normal distribution curve, allowing it to be plotted as a smooth line.

```python
# Creating evenly spaced values across the range of petal lengths.

x = np.linspace(petal_lengths.min(), petal_lengths.max(), 100)
```

```python
# Calculating the normal distribution using the petal length mean and standard deviation.

normal_distribution = stats.norm.pdf(x, petal_mean, petal_std)
```

### Petal Length Histogram and Normal Distribution

The normal distribution calculated using the mean and standard deviation of the petal length data is superimposed on the histogram. The histogram is displayed as a density so that it can be compared with the normal probability density curve.

```python
# Creating a histogram of petal lengths for all 150 flowers.

plt.hist(petal_lengths, density=True, color="forestgreen", edgecolor="black")
plt.plot(x, normal_distribution, color="red", linewidth=2, label="Normal Distribution")
plt.legend()
plt.xlabel("Petal Length (cm)")
plt.ylabel("Density")
plt.title("Distribution of Petal Lengths")
plt.show()
```
