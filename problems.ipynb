## Problem 1: Exploring Petal Length

Problem 1 investigates the distribution of petal length in the Iris dataset. It looks into the whole dataset and separate Iris species.

```python
# Importing libraries required for data analysis, visualisation and statistics.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

### Loading the Iris Dataset

The Iris dataset used for this analysis is loaded directly from the URL provided in the assessment brief. The dataset is read into a pandas DataFrame so that the petal length data can be analysed.

```python
# Loading the Iris dataset from the URL provided in the assessment brief.

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

iris = pd.read_csv(url)
iris.head()
```

### Iris Species

The Iris dataset contains three species: setosa, versicolor and virginica. Before comparing petal lengths between the species, the number of observations in each species is checked.

```python
# Counting the number of flowers belonging to each Iris species.

iris["species"].value_counts()
```

The dataset contains 50 observations for each of the three Iris species, giving 150 observations in total. Each species is equally represented in the dataset.

### Petal Length Distribution

The distribution of petal lengths for all 150 flowers is examined using a histogram.

```python
# Selecting the petal length values for all flowers.

petal_lengths = iris["petal_length"]
```

```python
# Creating a histogram of petal lengths for all 150 flowers.

plt.hist(petal_lengths)
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.title("Distribution of Petal Lengths")
plt.show()
```
