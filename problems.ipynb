### Preparing the Normal Distribution

To plot a normal distribution over the histogram, a range of values is needed for the x-axis. The `np.linspace()` function is used to create 100 evenly spaced values between the minimum and maximum petal lengths. These values will be used to calculate points along the normal distribution curve, allowing it to be plotted as a smooth line.

```python
# Creating evenly spaced values across the range of petal lengths.

x = np.linspace(petal_lengths.min(), petal_lengths.max(), 100)
```

```python
# Calculating the normal distribution using the petal length mean and standard deviation.

normal_distribution = stats.norm.pdf(x, petal_mean, petal_std)
```
